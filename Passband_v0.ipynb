```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters

lambda_min = 1033   # nm
lambda_max = 1047   # nm
num_lines  = 168
sigma      = 0.4       # nm 

# Derive
centers = np.linspace(lambda_min, lambda_max, num_lines)

# Wavelength grid for plotting
lambda_grid = np.linspace(lambda_min -3*sigma, lambda_max + 3*sigma, 20000)

# Gaussian function
def gaussian(x, mu, sigma):
    return np.exp(-(x - mu)**2 / (2 * sigma**2))

# Passband
passband = np.zeros_like(lambda_grid)

for mu in centers:
    passband += gaussian(lambda_grid, mu, sigma)

# Normalize
passband /= passband.max()

# Plot
plt.figure(figsize=(8, 4))
plt.plot(lambda_grid, passband, linestyle='-', linewidth='2', color='brown', alpha=0.9, visible=True)
plt.xlabel("Wavelength (nm)")
plt.ylabel("Normalized Spectral Response")
plt.title(f"168-Line Gaussian Passband (σ = {sigma:.3f} nm), peaks 0.06 nm apart")
plt.grid(True)
plt.tight_layout()
plt.axhline(y=0.707, linestyle='--', linewidth='1', color='green', alpha=0.7, visible=True, label='1/√2')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
lambda_min = 1034.99   # nm
lambda_max = 1045.01   # nm
num_lines  = 168
sigma      = 0.016       # nm  
db         = 3.01         # dB  (dynamic range)

# Parameters
centers = np.linspace(lambda_min, lambda_max, num_lines)

lambda_grid = np.linspace(lambda_min - 3*0.2,
                          lambda_max + 3*0.2,
                          2000)
# Gaussian 
def gaussian(x, mu, sigma):
    return np.exp(-(x - mu)**2 / (2 * sigma**2))

# Passband
passband = np.zeros_like(lambda_grid)

for mu in centers:
    passband += gaussian(lambda_grid, mu, sigma)

# Normalize
passband /= passband.max()

# Convert to dB and apply dynamic range
passband_db = 10 * np.log10(passband)
passband_db = np.clip(passband_db, -db, 0)

# Plot (dB scale)
plt.figure(figsize=(10, 4))
plt.plot(lambda_grid, passband_db)
plt.xlabel("Wavelength (nm)")
plt.ylabel("Spectral Response (dB)")
plt.title(f"168-Line Gaussian Passband (σ = {sigma:.3f} nm, DR = {db} dB)")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import erf

# Given pulse width
fwhm_ns = 2.0  # ns

# Gaussian conversions
sigma = fwhm_ns / (2 * np.sqrt(2 * np.log(2)))
hwhm = fwhm_ns / 2
width_1e2 = 4 * sigma

# Time axis
t = np.linspace(-5, 5, 2000)  # ns

# Gaussian pulse (normalized area = 1)
G = (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp(-t**2 / (2 * sigma**2))
peak = G.max()

# Levels
lvl_1 = peak * np.exp(-0.5)
lvl_2 = peak * np.exp(-2.0)
lvl_3 = peak * np.exp(-4.5)
lvl_half = 0.5 * peak

# Areas
area_1 = erf(1 / np.sqrt(2))
area_2 = erf(2 / np.sqrt(2))
area_3 = erf(3 / np.sqrt(2))
area_fwhm = erf(np.sqrt(np.log(2)))

# Plot
plt.figure(figsize=(12, 6))
plt.plot(t, G, color="black", lw=2, label="Gaussian Pulse")

# Vertical markers
for x, ls, c, lbl in [
    ( sigma, "--", "red", "±1σ"),
    (-sigma, "--", "red", None),
    ( 2*sigma, "--", "blue", "±2σ"),
    (-2*sigma, "--", "blue", None),
    ( 3*sigma, ":", "green", "±3σ"),
    (-3*sigma, ":", "green", None),
    ( hwhm, "-.", "orange", "±FWHM/2"),
    (-hwhm, "-.", "orange", None),
]:
    plt.axvline(x, linestyle=ls, color=c, label=lbl)

# Horizontal markers
plt.axhline(lvl_1, linestyle="--", color="red")
plt.axhline(lvl_2, linestyle="--", color="blue")
plt.axhline(lvl_3, linestyle=":", color="green")
plt.axhline(lvl_half, linestyle="-.", color="orange", label="Half Maximum")

# Text
plt.text(0.02, 0.92, f"FWHM = {fwhm_ns:.2f} ns", transform=plt.gca().transAxes)
plt.text(0.02, 0.87, f"σ = {sigma:.3f} ns", transform=plt.gca().transAxes)
plt.text(0.02, 0.82, f"1/e² width = {width_1e2:.2f} ns", transform=plt.gca().transAxes)

plt.text(0.02, 0.72, f"Area ±σ     : {area_1:.4f}", transform=plt.gca().transAxes)
plt.text(0.02, 0.67, f"Area ±2σ    : {area_2:.4f}", transform=plt.gca().transAxes)
plt.text(0.02, 0.62, f"Area ±3σ    : {area_3:.4f}", transform=plt.gca().transAxes)
plt.text(0.02, 0.57, f"Area ±FWHM/2: {area_fwhm:.4f}", transform=plt.gca().transAxes)

plt.xlabel("Time (ns)")
plt.ylabel("G(t)")
plt.title("1D Gaussian Pulse (FWHM = 2 ns)")
plt.legend()
plt.grid()
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import math
import matplotlib.pyplot as plt

# Constants
c = 299792458          # speed of light in m/s
SC_speed= 7102.493     # ground speed of SC  in m/s 
GSD_track=6            # along the track in m
Swath_width=1008       # in m. This so that 3 sweeps onto 56 pixels can give us 168 spots on the cross track
Cross_track_spots=168  # Number spots on the cross track

# Input Parameters
grating_lpm = 1503.76                  # lines per mm
center_wavelength_nm = 1040            # central wavelength (nm)
wavelength_span_nm = 10.02             # total span (nm)
altitude_km = 474                      # observation altitude (km)

# Derived Values
d = 1 / (grating_lpm * 1e3)            # groove spacing in meters
lambda_center = center_wavelength_nm * 1e-9
delta_lambda = wavelength_span_nm * 1e-9
altitude_m = altitude_km * 1e3

# Angular shift (small angle approx)
delta_theta_rad = Swath_width / altitude_m

# The effective angular separation between each wavelength :
ang_sep=delta_theta_rad/(Cross_track_spots-1)

# Ground footprint
footprint_m = altitude_m * delta_theta_rad
footprint_km = footprint_m / 1000

# Frequency band for 1040±5.01 nm
span_freq_band=(c/1034.99e-9)-(c/1045.01e-9)

# Frequency between each subsequent beam assuming 168 spots per swath relative to 1040±5.01 nm
span_freq_spot = span_freq_band/(Cross_track_spots-1)

# RepRate assuming 256 cross tracks with using GSD_plane above
RepRate=(SC_speed/GSD_track)*Cross_track_spots

# Print summary
print("--- Calculations for 474 km ---")
print(f"Grating:                      {grating_lpm} lines/mm")
print(f"Wavelength span:              {wavelength_span_nm} nm around {center_wavelength_nm} nm")
print(f"Altitude:                     {altitude_km} km")
print(f"Angular spread:               {delta_theta_rad*1e3:.3f} mrad or {np.degrees(delta_theta_rad):.2f} degrees")
print(f"Angular separation b/w beams: {ang_sep*1e6:.3f} µrad or {np.degrees(ang_sep):.5f} degrees")
print(f"Beam footprint on ground:     {footprint_km*1e3:.1f} m")
print(f"Freq span band:               {span_freq_band/1e9:.3f} GHz")
print(f"Freq span for each spot:      {span_freq_spot/1e9:.3f} GHz")
print(f"RepRate:                      {RepRate/1e3:.3f} kHz")
```
