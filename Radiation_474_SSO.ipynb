```python
# Simple exploratory model: dose vs Al thickness for multiple orbit altitudes
# - Toy model only: use AE9/AP9 or HZETRN for engineering-level results.
# - Requires: numpy, matplotlib

import numpy as np
import matplotlib.pyplot as plt

# --- User parameters (adjustable) ---
thickness_mm = np.linspace(0.05, 20.0, 200)   # aluminum thickness in mm
# convert mm -> g/cm^2: density_Al(g/cm3)=2.70 ; thickness (cm)=mm/10
def mm_to_gcm2(mm):
    return 2.70 * (mm / 10.0)

mass_depth = mm_to_gcm2(thickness_mm)  # g/cm^2

# Altitudes to examine (km)
altitudes_km = [200, 400, 600, 1000, 2000, 6000, 10000, 20000, 35786]
alt_labels = [f"{a} km" for a in altitudes_km]

# --- Toy unshielded dose model (arbitrary units) ---
# Components: constant GCR baseline + two Gaussian trapped-belt peaks (inner, outer)
def unshielded_dose(alt_km):
    # baseline GCR (slowly varying with altitude)
    gcr_base = 1.0  # arbitrary units
    # inner belt (rough center, not exact) - inner-belt-like contribution
    inner_amp = 20.0
    inner_center = 2000.0
    inner_sigma = 1500.0
    inner = inner_amp * np.exp(-0.5 * ((alt_km - inner_center)/inner_sigma)**2)
    # outer belt (again, toy)
    outer_amp = 10.0
    outer_center = 10000.0
    outer_sigma = 3000.0
    outer = outer_amp * np.exp(-0.5 * ((alt_km - outer_center)/outer_sigma)**2)
    # South Atlantic Anomaly bump at low altitudes (rough)
    saa_amp = 3.0 * np.exp(-0.5 * ((alt_km - 500.0)/300.0)**2)
    return gcr_base + inner + outer + saa_amp

# evaluate unshielded doses for altitudes
D0_list = [unshielded_dose(a) for a in altitudes_km]

# --- Shielding model ---
# Exponential attenuation of primaries + simple "buildup" term to model secondaries.
# D_shield = D0 * exp(-mu_eff * mass_depth) + D0 * secondary_frac * (1 - exp(-mass_depth / mass_scale))
# where:
# - mu_eff [1/(g/cm^2)] is an effective attenuation coefficient (tunable)
# - secondary_frac is fraction of D0 that can reappear as secondaries at large mass depths
# - mass_scale [g/cm^2] sets how quickly buildup approaches its asymptote
mu_eff = 0.06   # 1/(g/cm^2)  -- chosen for exploratory behavior; replace with energy-specific mu/ρ from NIST for more fidelity.
secondary_frac = 0.25
mass_scale = 30.0  # g/cm^2 (rough scale for secondary buildup)

def shielded_dose(D0, mass_depth_gcm2):
    primaries = D0 * np.exp(-mu_eff * mass_depth_gcm2)
    secondaries = D0 * secondary_frac * (1.0 - np.exp(-mass_depth_gcm2 / mass_scale))
    return primaries + secondaries

# compute shielded doses
dose_curves = []
for D0 in D0_list:
    dose_curves.append(shielded_dose(D0, mass_depth))

# --- Power-law overlay (t^-2) normalized to each D0 at t = 1 mm for comparison ---
power_law_ref = {}
for i, D0 in enumerate(D0_list):
    # avoid division by zero: use thickness_mm >= 0.1
    c = D0 * (1.0) ** 2  # normalization so that at 1 mm the curve equals D0 (chooseable)
    power_law_ref[alt_labels[i]] = c * (thickness_mm ** -2.0)

# --- Plotting ---
plt.figure(figsize=(10, 7))
for i, label in enumerate(alt_labels):
    plt.loglog(thickness_mm, dose_curves[i], label=label, linewidth=1.6)

# plot power-law (example for mid-altitude)
plt.loglog(thickness_mm, power_law_ref[alt_labels[3]], '--', label='power-law t^-2 (norm @1mm)', linewidth=1.6)

plt.xlabel("Aluminum thickness (mm)")
plt.ylabel("Dose (arbitrary units)")
plt.title("Toy model: Dose vs Aluminum thickness for different orbit altitudes\n(attenuation + simple secondary buildup)")
plt.grid(which='both', linestyle=':', linewidth=0.5)
plt.legend(ncol=2, fontsize=9)
plt.xlim(0.05, 20)
plt.ylim(1e-2, 1e4)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Example data (mm, krad(Si))
data = {
    'Al_thickness_mm': [0.05, 0.4, 1, 2, 5, 10],
    'Total_dose_krad': [150, 80, 40, 25, 18, 15]
}

df = pd.DataFrame(data)

plt.loglog(df['Al_thickness_mm'], df['Total_dose_krad'], 'o-', label='SPENVIS data')
fit = np.polyfit(np.log(df['Al_thickness_mm']), np.log(df['Total_dose_krad']), 1)
plt.loglog(df['Al_thickness_mm'], np.exp(np.polyval(fit, np.log(df['Al_thickness_mm']))),
           '--', label=f'Power law: t^{fit[0]:.2f}')

plt.xlabel('Aluminum thickness (mm)')
plt.ylabel('Total Dose [krad(Si)]')
plt.title('SPENVIS: Total Dose vs Aluminum Shielding')
plt.legend()
plt.grid(True, which='both')
plt.show()
```
