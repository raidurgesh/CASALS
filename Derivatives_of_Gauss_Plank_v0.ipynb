```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
sigma = 2.0
x = np.linspace(-10, 10, 2000)
dx = x[1] - x[0]

# Gaussian
G = (1 / (np.sqrt(2 * np.pi) * sigma)) * np.exp(-x**2 / (2 * sigma**2))

# First derivative
dG = -(x / sigma**2) * G

# Second derivative
ddG = ((x**2 - sigma**2) / sigma**4) * G

# L2 normalization
def normalize(f, dx):
    return f / np.sqrt(np.sum(f**2) * dx)

G_n = normalize(G, dx)
dG_n = normalize(dG, dx)
ddG_n = normalize(ddG, dx)

# Plots
plt.figure(figsize=(10, 6))
plt.plot(x, G_n, label="Normalized Gaussian")
plt.plot(x, dG_n, label="Normalized 1st Derivative")
plt.plot(x, ddG_n, label="Normalized 2nd Derivative")
plt.xlabel("x")
plt.ylabel("Amplitude (Normalized)")
plt.title("1D Gaussian and Its Normalized Derivatives")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Physical constants
h = 6.62607015e-34      # Planck constant [J*s]
c = 2.99792458e8        # Speed of light [m/s]
kB = 1.380649e-23       # Boltzmann constant [J/K]

# Parameters
T = 5778  # Sun temperature [K]
lam = np.linspace(200e-9, 2000e-9, 3000)  # wavelength [m]
dlam = lam[1] - lam[0]

# Planck spectral radiance (per wavelength)
def planck_lambda(lam, T):
    return (2*h*c**2 / lam**5) / (np.exp(h*c/(lam*kB*T)) - 1)

B = planck_lambda(lam, T)

# Numerical derivatives
dB = np.gradient(B, dlam)
ddB = np.gradient(dB, dlam)

# L2 normalization
def normalize(f, dx):
    return f / np.sqrt(np.sum(f**2) * dx)

B_n = normalize(B, dlam)
dB_n = normalize(dB, dlam)
ddB_n = normalize(ddB, dlam)

# Plot
plt.figure(figsize=(10, 6))
plt.plot(lam * 1e9, B_n, label="Normalized Planck Radiance")
plt.plot(lam * 1e9, dB_n,  label="Normalized 1st Derivative")
plt.plot(lam * 1e9, ddB_n, label="Normalized 2nd Derivative")
plt.xlabel("Wavelength (nm)")
plt.ylabel("Normalized Amplitude")
plt.title("Planck Spectral Radiance and Its Normalized Derivatives (T = 5778 K)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
