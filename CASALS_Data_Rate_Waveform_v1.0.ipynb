```python
###### Waveform CASALS Data Rate Calculation

# Input Parameters
PRF = 197353                        # Reprate
samples_per_waveform_Rx = 200       #(200 for Rx)
samples_per_waveform_Tx = 20        #(20 for Tx,need to capture sent out and pulse shape accounting for jitter)
samples_per_waveform_Start = 20     #(20 for start pulse)
samples_per_waveform_Overhead = 20  # (Overhead GPS signal)
bits_per_sample = 16                # Bits per sample
num_digitizer = 2                   # Number of Digitizers
#num_channels = 64                  # Number of Channels (This is not considered for evaluations since Reprate captures the number of outputs)
compression_factor = 1              # Compression Ratio (2 or can be upto 3)
atm_background_binning_points = 26  # 26 for atm background, pulse window of 5098 sec, light travels 1528m, so 1528/26 - 59m
bits_per_background_point = 16      # Bits per sample

# Waveform pulse data rate
samples_per_pulse = samples_per_waveform_Tx + samples_per_waveform_Rx + samples_per_waveform_Start
data_rate_bps = PRF * samples_per_pulse * bits_per_sample * num_digitizer / compression_factor
data_rate_MBps = data_rate_bps / (8 * 10**6)  # Convert bits per second to Megabytes/s

# Atmospheric background data rate
data_rate_atm_background_bps = PRF * atm_background_binning_points * bits_per_background_point / compression_factor
data_rate_atm_background_MBps = data_rate_atm_background_bps / (8 * 10**6)

# Display Results
print("\n=== Waveform CASALS Data Rate Results ===")
print(f"Data Rate (Waveform): {data_rate_bps:.2e} bits/s ({data_rate_MBps:.2f} MBPS)")
print(f"Data Rate (Background): {data_rate_atm_background_bps:.2e} bits/s ({data_rate_atm_background_MBps:.2f} MB/s)")
```
